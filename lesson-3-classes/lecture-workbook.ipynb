**Quizlet: property or method?** `mass` below is a `@property`.
One of these two lines works: `z.mass` or `z.mass()`. Which one, and what error does the other give?

```python
class FourVector:
    def __init__(self, E, px, py, pz):
        self.E = E
        self.px = px
        self.py = py
        self.pz = pz

    @property
    def mass(self):
        return np.sqrt(self.E**2 - self.px**2 - self.py**2 - self.pz**2)

z = FourVector(91.2, 0.0, 0.0, 0.0)
```

**Quizlet: inheritance.** `TightSelection` defines no `apply` of its own.
Before you run it: what does the last line give, and whose code ran?

```python
class Selection:
    def __init__(self, min_pt):
        self.min_pt = min_pt

    def apply(self, pt):
        return pt > self.min_pt

class TightSelection(Selection):
    def __init__(self):
        super().__init__(min_pt=50)

TightSelection().apply(np.array([63.0, 38.1, 4.0]))
```
