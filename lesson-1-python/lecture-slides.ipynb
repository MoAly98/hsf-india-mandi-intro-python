## What the detector sees

We see only the leptons. Their summed four-momentum gives back the mass of the parent.

$$m = \sqrt{E^2 - p_x^2 - p_y^2 - p_z^2}$$

## What we look for

A particle that decays is never seen directly: the detector records only its decay products. Energy and momentum are conserved in the decay, so the four-momenta of the products add up to the four-momentum of the parent, and $m = \sqrt{E^2 - p^2}$ applied to that sum gives back the parent's mass.

In a real event we do not know which particles came from the same decay. Combining the right products returns the parent's mass every time; combining unrelated particles returns a value with no special meaning. A histogram of the mass of every combination in every event therefore shows a smooth spread from the wrong combinations, with a peak at the parent's mass from the right ones.

A peak at a mass where none was expected is how a new particle is discovered. Finding the Higgs peak at 125 GeV with Python's analysis tools is the main objective of this course.

## Pairs

A loop inside a loop can be used to collect every pair. `i < j` keeps each pair once.

`range(n)` counts `0, 1, ..., n-1`. 

`(i, j)` is a *tuple*: similar to a list but cannot be modified once created (immutable)

<img src="../img/pairs-loop.svg" width="70%" alt="">

````python
pts = [63.0, 38.1, 4.0]
pairs = []
for i in range(len(pts)):
    for j in range(len(pts)):
        if i < j:
            pairs.append((i, j))
pairs
````

## `enumerate` and `zip`

`enumerate` loops over the index of a list and its values at the same time

````python
for i, pt in enumerate([63.0, 38.1]):
    print(i, pt)
````

`zip` walks two lists at the same time

````python
for pt, eta in zip([63.0, 38.1], [-0.72, -0.88]):
    print(pt, eta)
````

## A little data analysis

Starting from the observed electrons and muons, we reconstruct unobserved particles by adding energy and momentum.

<img src="../img/higgs-to-four-leptons-diagram.png" width="50%" alt="">

````python
particles = [
    {
        "type": "electron",
        "E": 171.848714,
        "px": 38.4242935,
        "py": -28.779644,
        "pz": 165.006927,
        "charge": 1,
    },
    {
        "type": "electron",
        "E": 138.501266,
        "px": -34.431419,
        "py": 24.6730384,
        "pz": 131.864776,
        "charge": -1,
    },
    {
        "type": "muon",
        "E": 68.1289790,
        "px": -17.945541,
        "py": 13.1652603,
        "pz": 64.3908386,
        "charge": 1,
    },
    {
        "type": "muon",
        "E": 18.8320473,
        "px": -8.1843795,
        "py": -7.6400470,
        "pz": 15.1420097,
        "charge": -1,
    },
]
````

````python
def particle_decay(name, particle1, particle2):
    return {
        "type": name,
        "E": particle1["E"] + particle2["E"],
        "px": particle1["px"] + particle2["px"],
        "py": particle1["py"] + particle2["py"],
        "pz": particle1["pz"] + particle2["pz"],
        "charge": particle1["charge"] + particle2["charge"],
    }
````

````python
z1 = particle_decay("Z boson", particles[0], particles[1])
z1
````

````python
z2 = particle_decay("Z boson", particles[2], particles[3])
z2
````

````python
higgs = particle_decay("Higgs boson", z1, z2)
higgs
````
