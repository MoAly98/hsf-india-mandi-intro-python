## What the detector sees

We see only the leptons. Their summed four-momentum gives back the mass of the parent.

$$m = \sqrt{E^2 - p_x^2 - p_y^2 - p_z^2}$$

## Why a peak means a particle

Add up the four-momenta of the right decay products and the parent's mass comes back.
Add up an unrelated set and you get a random number.

Do this in every event and fill a histogram: the random combinations spread out, the true decays pile up at the parent's mass. A particle is a bump.

This is how the Z was found, and the Higgs. It is what you will do all four lessons.

## Loops and `if`

A `for` loop runs the indented block once per item. `if` runs it only when a condition is `True`.

<img src="../img/for-loop.svg" width="48%" alt=""> <img src="../img/if.svg" width="48%" alt="">


````python
for electron in event["electron"]:
    if electron["pt"] > 45:
        print(electron["pt"])
````

Python runs one statement at a time and never looks ahead: trouble on the next line does not stop the current one. `elif` and `else` pick between blocks:

````python
for electron in event["electron"]:
    if electron["pt"] > 45:
        print("leading")
    elif electron["pt"] > 20:
        print("good")
    else:
        print("soft")
````

## Selections

Loop, test, collect. This pattern is a selection.

<img src="../img/filter.svg" width="70%" alt="">

````python
good = []
for electron in event["electron"]:
    if electron["pt"] > 45:
        good.append(electron)
len(good)
````

## Comprehensions

Loop, test, collect is so common that Python has a one-line form. It reads like the sentence: *the pT of every electron above 45*.

<img src="../img/comprehension.svg" width="80%" alt="">


````python
good = [e["pt"] for e in event["electron"] if e["pt"] > 45]
good
````

The same form builds dictionaries:

````python
{field: 0.0 for field in ["E", "px", "py", "pz"]}
````

## Pairs

A loop inside a loop visits every pair. `i < j` keeps each pair once.

`range(n)` counts 0, 1, ..., n-1. `(i, j)` is a tuple: a short list that cannot change.

<img src="../img/combinations.svg" width="70%" alt="">

````python
pairs = []
for i in range(len(pts)):
    for j in range(len(pts)):
        if i < j:
            pairs.append((i, j))
pairs
````

## Functions and `import`

A function gives a name to a formula. `import` brings in code that other people wrote.

<img src="../img/function.svg" width="70%" alt="">

````python
import math

def transverse(px, py):
    return math.sqrt(px**2 + py**2)

transverse(3.0, 4.0)
````

A parameter can carry a default. Give it only when yours differs:

````python
def passes(pt, min_pt=20.0):
    return pt > min_pt

passes(63.0), passes(63.0, min_pt=100.0)
````

Indent with spaces, 4 per level, never the tab key: tabs and spaces look identical and do not mix.

````python
shorter = transverse
shorter(3.0, 4.0)
````

`import math` brings in the whole library, and the dot keeps its names from colliding with yours. To take just one thing:

## Puzzle: how fast is a muon?

A muon flies through CMS with an energy of 63 GeV. Its mass is 0.106 GeV.

Relativity gives the recipe: $\gamma = E/m$, then $\beta = \sqrt{1 - 1/\gamma^2}$ is the speed as a fraction of the speed of light, 299{,}792 km/s.

How many km/s slower than light is it? Guess an order of magnitude first.

<img src="../img/muon-speed.svg" width="80%" alt="">


````python
gamma = 63.0 / 0.106
beta = math.sqrt(1 - 1 / gamma**2)
(1 - beta) * 299792.458
````

Less than half a kilometre per second slower than light. That is why every muon here gets `E` and momentum, and no speed: they are all just below c.
